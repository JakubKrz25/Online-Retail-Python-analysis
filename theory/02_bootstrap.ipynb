### To derive
- Bootstrap principle: approximate the law of $\sqrt n\,(T(\hat F_n) - T(F))$ under $F$ by the law of $\sqrt n\,(T(\hat F_n^*) - T(\hat F_n))$ under $\hat F_n$.
- Glivenko–Cantelli and the DKW inequality; Hadamard differentiability and the functional delta method (why the bootstrap works for smooth $T$).
- Percentile, basic and BCa intervals: where each one comes from.
- A case where the bootstrap fails: the sample maximum, $T(F) = \sup \operatorname{supp} F$.

- Same for the median and for a "top 10% share" statistic.
- The bootstrap failure for the maximum of $U(0, \theta)$.
