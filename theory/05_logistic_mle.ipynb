### To derive
- Model $P(Y=1 \mid x) = \sigma(x^\top\beta)$; log-likelihood $\ell(\beta) = \sum_i \big[y_i x_i^\top\beta - \log(1 + e^{x_i^\top\beta})\big]$.
- Gradient and Hessian; concavity of $\ell$; Newton–Raphson = IRLS.
- Negative log-likelihood = binary cross-entropy (the link to Raschka / deep learning).
- Asymptotic normality of the MLE, Fisher information, Wald CIs for odds ratios.
- Separation: when the MLE does not exist.
